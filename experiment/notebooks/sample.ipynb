```python
import sys
sys.path.append('/work/pi_huiguan_umass_edu/sandeep/code/dgl_groot/python')
scratch_dir = '/scratch/workspace/spolisetty_umass_edu-graph_data/'
import torch
import torch.distributed as dist
from dgl.dev.dataloader import *
import dgl
device = torch.cuda.device(0)
```

```python
from ogb.nodeproppred import DglNodePropPredDataset
dataset = DglNodePropPredDataset(name="ogbn-papers100M", root = scratch_dir)
split_edge = dataset.get_idx_split()
graph, labels = dataset[0]
graph = graph.to(0).formats('csc')
train = split_edge['train'].to(0)
```

```python
# train_nid = th.nonzero(g.ndata["train_mask"], as_tuple=True)[0]

# Create PyTorch DataLoader for constructing blocks
sampler = dgl.dataloading.MultiLayerNeighborSampler(
    [10,10,10,10,10,10]
)
dataloader = dgl.dataloading.DataLoader(
    graph,
    train,
    sampler,
    batch_size=1024,
    shuffle=True,
    drop_last=False,
)
```

```python
for batch_data  in dataloader:
    src,dest, blocks = batch_data
    print(src.shape[0], blocks[0].num_edges())
```
